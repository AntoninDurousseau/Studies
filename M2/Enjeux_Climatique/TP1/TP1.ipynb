```python
import geopandas as gpd
from pathlib import Path
```

```python
data = r"C:\Users\Antonin\Studies\M2\Enjeux_Climatique\BDTOPO_3-5_TOUSTHEMES_SHP_LAMB93_D075_2026-09-15\BDTOPO\1_DONNEES_LIVRAISON_2026-09-00210\BDT_3-5_SHP_LAMB93_D075_ED2026-09-15"

commune = gpd.read_file(data + r"\ADMINISTRATIF\Commune.shp")
batiment = gpd.read_file(data + r"\BATI\BATIMENT.shp")
terrain = gpd.read_file(data + r"\BATI\TERRAIN_DE_SPORT.shp")
seine = gpd.read_file(data + r"\HYDROGRAPHIE\COURS_D_EAU.shp")
transport = gpd.read_file(data + r"\TRANSPORT\EQUIPEMENT_DE_TRANSPORT.shp")
voie = gpd.read_file(data + r"\TRANSPORT\TRONCON_DE_VOIE_FERREE.shp")
parc = gpd.read_file(data + r"\OCCUPATION_DU_SOL\ZONE_DE_VEGETATION.shp")
```

```python
paris = commune[commune['NOM'] == 'Paris']
metro = transport[transport['NATURE'] == 'Station de métro']
residence = batiment[batiment['USAGE1'] == 'Résidentiel']
```

```python
residence_paris = gpd.sjoin(residence, paris, how="inner", predicate="within")
metro_paris = gpd.sjoin(metro, paris, how="inner", predicate="within")
terrain_paris = gpd.sjoin(terrain, paris, how="inner", predicate="within")
parc_paris = gpd.sjoin(parc, paris, how="inner", predicate="within")
```

```python
zone_eau = seine.buffer(100).union_all()
zone_metro = metro_paris.buffer(250).union_all()
zone_parc = parc_paris.buffer(100).union_all()
zone_terrain = terrain_paris.buffer(750).union_all()
```

```python
habitation_aggreable = residence_paris[
    residence_paris.within(zone_metro) &
    residence_paris.within(zone_parc) &
    residence_paris.within(zone_eau) &
    residence_paris.within(zone_terrain)
]

print("Nombre d'habitations agréables :", len(habitation_aggreable))
```
